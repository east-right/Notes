# 해시테이블 - 체이닝방식 구현

```python
# HashTable-chainning에 들어갈 Node
class HashNode:
    def __init__(self, key, val, next = None):
        self.key = key
        self.val = val
        self.next = None
```

```python
class HashTable:
    def __init__(self): #기본적으로 HashTable 구성에 필요한 정보를 적는다.
        self.size = 10 # 테이블의 길이를 지정한다.
        self.table =[None] * self.size # 헤시테이블을 하나 만든다. 헤시테이블의 크기를 설정을 하여 Bucket이 최대 몇개인지 설정한다.
    
    def _hash_function(self, key): # Key를 받으면 특정 해시를 반환할 함수를 만들어야 한다. // 여기서는 size의 나머지를 만들어 주는 해시함수를 만들것 이다.
         return key % self.size
    
    def put(self, key, val): #삽입
        idx = self._hash_function(key)#size를 나눈 나머지를 값으로 반환하는 해시함수에 Key값을 넣어야한다. // 해시값(idx)을 찾는 코드이다.
        
        if self.table[idx] is None: # 해시값에 해당하는 테이블의 위치가 비어있는지 물어보는 코드이다.
            self.table[idx] = HashNode(key, val, None) # 해당 테이블에 값이 비어있으면 그냥 넣으면 된다. 쉽다. // next 자리에 굳이 None을 해주지 않아도 되지만 가독성을 위해
        
        else:# 값이 있는 경우는 Queue 방식을 통해서 LinkedList를 만든다. 
            node = self.table[idx] # 기존에 있는값을 불러온다. // 저 Node 비어있는 next에 값을 넣어야 하니까.
            while node.next is not None: #Slot이라는 LinkedList의 맨 뒤로 이동하는 함수// next가 None이 나올 때까지 
                node = node.next # 계속 다음 Node를 불러온다.
            
            node.next = HashNode(key, val, None) #While 문을 통해서 기존에 존재하던 버킷의 node중 젤 뒤로 왔으니 거기에 새로운 Hashnode를 집어 넣는다.
    def get(self, key): # 조회
        idx = self._hash_function(key) #먼저 해시값을 구한다.
        node = self.table[idx] #테이블의 초기값 즉 head뽑아본다.
        
        while node is not None: #node가 비어있지 않을 때 아래 코드를 실행한다.
            if node.key == key: # node의 key가 내가 찾고싶어하는 key랑 맞으면
                return node.val # val 값을 보여준다.
            node = node.next # if에 걸리지 않으면 즉 내가 찾고 싶어하는게 아니면 다음 노드를 확인한다.
        return -1 # 이건 while문에 걸렸을 때 즉 node가 비어있다는 뜻인 node=None일 때 -1을 반환하게 한다. None이면 while 아래에 있는 코드를 실행 및 반복을 하지 않기에 바로 return으로 내려온다.
    
    def remove(self, key): #삭제
        idx = self._hash_function(key) #먼저 해시값을 구한다.
        node = self.table[idx] #테이블의 초기값 즉 head뽑아본다.
        prev_node = None #이건 삭제를 위한 이전 노드 변수이다. // 사용자가 key에 해당하는 node를 삭제시키고 싶어하면 그냥 key 기준의 앞의 node와 뒤어 노드를 연결시키면 되니까
        
        while node is not None: #만약 node가 비어있지 않으면
            if node.key == key: # node의 key가 내가 찾고싶어하는 key랑 맞으면
                if prev_node is not None: # 프리뷰노드 즉 지금 노드의 앞의 노드가 존재하면  // 삭제시키고자 하는 key값이 맨 처음 즉 head가 아니라면 
                    prev_node.next = node.next # 이전노드의 방향을 다음노드로 설정을 해서 지금 노드는 뒷방늙은이로 만든다.
                else: # 반약 prev_node가 비어있다 == 이게 지금 이 버켓에 존재하는 LinkedList의 head라는 거니깐
                    self.table[idx] = node.next # head의 바로 다음 값을 지금 bucket에 넣어주면 지금 노드는 뒷방늙은이가 된다.
                return #키값을 찾아 작업을 완료했으니 끝낸다.
            #여기는 반복문 중에 key 매칭을 찾는 if에 걸리지 않았을 때 // 즉 내가 보고있는 해당 bucket의 slot이 내가 찾는 위치가 아니면
            prev_node = node #지금 값은 다음 값의 이전이 되니깐 저장 
            node = node.next # 다음거 찾아보러 고고싱~
```

```python
"""
아래와 같이 None으로 일단 값이 없다는걸 지정을 해주면 메모리는 사용하지만 비어있다는걸 명시적으로 표현이 가능하다. 
내가 None으로 빈값을 저장하는게 익숙하지 않다. 하지만 이번 자료구조를 통해서 결측치 혹은 빈값을 None으로 처리하는 방법을 안거 같아서 좋다.
이 방법을 다음에도 내가 잘 사용해보자자
"""
print([None] *10)
print([] *10)
```

```python
ht = HashTable()

ht.put(1, 1)
ht.put(2, 2)
assert ht.get(1) == 1
assert ht.get(2) == 2
assert ht.get(3) == -1

ht.put(12, 1)
ht.put(22, 2)
ht.put(32, 3)
assert ht.get(12) == 1
assert ht.get(22) == 2
assert ht.get(32) == 3

ht.remove(12)
assert ht.get(2) == 2
assert ht.get(12) == -1
assert ht.get(22) == 2
assert ht.get(32) == 3

ht.get(2)
```
