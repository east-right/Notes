# Stack 기초

````python
class Node:
    def __init__(self, val, next): #기본적으로 Node에는 값(Value)와 다음 위치(next)를 들고 있어야 한다.
         self.val = val
         self.next = next
````

````python
class Stack:
    def __init__(self):
        self.top = None #초기값은 당연이 아무것도 없어서 None임
        
    #넣기(값 채우기기)
    def push(self, val):
        node = Node(val, None) #신규 Node가 들어옴
        node.next = self.top #신규 노드는 자기보다 먼저 들어온값(기존의 top)을 가르켜야함
        self.top = node
        #self.top = Node(val, self.top) #이 코드로 간단하게 구현가능
        #
    
    #빼기(값 확인하기, stack을 확인하면 값이 빠짐 빼야하니깐깐)
    def pop(self):
        if not self.top:
            return None  # top이 비어있다. 뺄게 없으니 None 배출
        
        node = self.top # 가장 위에 있는 값을 보여 줘야하니 top이 node가 됨
        self.top = node.next # 가장 높은 값을 보여줬으니 바로 아랫값이 top이 된다.
        return node.val
        
        
    #Stack이 비어있냐고 물어보는 함수
    def is_empty(self):
        return self.top is None # 비어있으면 True, 비어있지 않으면 False => 이게 왜 이렇게 되면 is가 == 이기에, if문 써야 되는거 아님? 이라고 순간 생각했는데 생각해보니 if문이 True, False에 반응하는거
````

````python
def test_stack():
    stack = Stack()

    stack.push(1) # 초기값 node(1, None)
    stack.push(2) # node(2, node(1, None))
    stack.push(3) # node(3, node(2, node(1, None)))
    stack.push(4) # node(4, node(3,node(2, node(1, None))))
    stack.push(5) # node(5, node(4, node(3,node(2, node(1, None)))))

    assert stack.pop() == 5 # 5를 반환, 실행 후 상태:node(4, node(3,node(2, node(1, None))))
    assert stack.pop() == 4 # 4를 반환, 실행 후 상태 node(3, node(2, node(1, None)))
    assert stack.pop() == 3 # 3을 반환, 실행 후 상태 node(2, node(1, None))
    assert stack.pop() == 2 # 2를 반환, 실행 후 상태 node(1, None)
    assert stack.pop() == 1 # 1을 반환, 실행 후 상태 None
    assert stack.pop() is None # None을 반환, 실행 후 상태 None
    assert stack.is_empty() # 비어있냐? ㅇㅇㅇ
````

````python
test_stack() # 이게 아무 결과 값이 안나오면 오류 없이 잘 돌아간다는 뜻이다.
````

# 유효한 괄호 문제

- 이 문제의 핵심은 괄호가 제대로 닫혀있나 이다.
- 팰린들롬 문제랑 비슷하다. 다른거라고 한다면 고려애햐하는 짝이 괄호들 뿐이라 쉬움
- 아래 코드들이 assert error가 안뜨게 돌아가는 함수 짜면 성공


```
assert test_problem_stack("()")
assert test_problem_stack("()[]{}")
assert test_problem_stack("({[][]})")
assert test_problem_stack("({[]})")
assert not test_problem_stack("(]")
assert not test_problem_stack("(()]")
assert not test_problem_stack("(((])")
assert not test_problem_stack("((())")
```

````python
def test_problem_stack(str):
    dic = {
        '}':'{',
        ')':'(',
        ']':'['
    }
    opener = '({[' #이건 괄호문제이기에 괄호는 무조건 저것들 중 하나가 먼저 나옴
    stack = Stack() #스텍 선언
    
    for char in str: #문자열 하나하나 잘라서 넣는다.
        if char in opener: # 만약 {, [, ( 중 하나다?
            stack.push(char) # 스텍에 밀어 넣어라
        
        else: # 만약 }, ], ) 중 하나다?
            if not stack: # stack이 비어있다? 자매품: if stack.is_empty()
                return False
            
            top = stack.pop() # 나중에 들간거 부터 빼라
            if dic[char] != top: # 그러고 dic으로 매칭하여 요놈의 짝을 찾고 top과 짝이 같지 않다?
                return False # False 실패 반환
    
    return stack.is_empty() # 실패 반환없이 stack이 None이다? 즉 비어있다? True를 반환하며 성공을 알려라
````

````python
"""
위에 방법은 Stack 형식으로 문제를 푼거고 Python에선 그냥 List가 짱이다. ㅋㅋ
"""
def test_problem_stack(s):
    pair = {
        '}': '{',
        ')': '(',
        ']': '[',
    }
    opener = "({["
    stack = []

    for char in s:
        if char in opener:
            stack.append(char) # stack.push(char)
        else:
            if not stack:
                return False
            top = stack.pop() # stack.pop()
            if pair[char] != top:
                return False

    return not stack
````

````python
test_problem_stack('{}')
````

````python
assert test_problem_stack("()")
assert test_problem_stack("()[]{}")
assert test_problem_stack("({[][]})")
assert test_problem_stack("({[]})")
assert not test_problem_stack("(]")
assert not test_problem_stack("(()]")
assert not test_problem_stack("(((])")
assert not test_problem_stack("((())")
````
