- 만약 def __init__(self, val, next): 로 작성하면 호출시 val과 next에 값을 입력해주지 않으면 error가 뜬다.

```python
class ListNode:
    def __init__(self, val, next):
        self.val = val
        self.next = next

#ll = ListNode() #error
ll = ListNode(0, None)
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

ll = ListNode()
```

- LinkedList를 만드는 코드는 아래와 같다.
- 여기서 head는 LinkedList의 시작점이고 LinkedList를 탐색하기 위해선 무조건 head로 들어와서 노드의 방향을 쫓아서 탐색해야한다.

```python
class LinkedList:
    def __init__(self):
        self.head = None #LinkedList 선언(생성)때는 값이 없기에 None 처리

    def append(self, val): # value 값을 넣음
        if not self.head: # 만약 head가 비어있으면(None)이면(if 만 있으면 True일 때 아래코드 if not이라 False일 때 아래코드) 아래코드를 진행해라
            self.head = ListNode(val, None)# 이제 첫 값이 들어왔으니 노드에 최초값을 넣어주고 next는 아직 다른 node가 없으니 None이다. 
            return

        node = self.head #이제 삽입을 해야하니 먼저 시작점인 head 부터 들어간다.
        while node.next: #이제 head에서 시작해서 None이 나올때까지 반복한다.
            node = node.next #기존 node의 .next에 있는 값을 node에 넣는다. 즉 next에 있는 방향으로 계속 이동한다. 

        node.next = ListNode(val, None) #방향을 따라간 가장 뒤의 next(None)에 새로은 LinstNode를 넣는다.
```

```python
ln = LinkedList()
ln.append(1)
ln.append(2)
ln.append(3)
```

```python
ln.head.val # 첫 노드
```

```python
ln.head.next.val #두 번째 노드
```

```python
ln.head.next.next.val #세 번째 노드 
```

# 팰린드롬 연결리스틑 문제

주어진 리스트가 팰린드롬인지 판별하는 프로그램을 짜시오
- [1,2,2,1] -> True
- [1,2] -> False

이 문제를 LinkedList에 데이터가 들어있다 가정했을 때 짜보자

```python
l1 = LinkedList()
for i in [1,2,2,1]:
    l1.append(i)
    
l2 = LinkedList()
for i in [1,2]:
    l2.append(i)
```

```python
print(l1.head.val)
print(l1.head.next.val)
print(l1.head.next.next.val)
print(l1.head.next.next.next.val)
```

```python
def isPalindrome(ln):
    arr = []
    head = ln.head

    if not head: # 비어있으면 True로 반환
        return True

    node = head # 시작점 head를 node로
    while node:
        arr.append(node.val) # node의 valeu를 arr이란 어레이에 저장
        node = node.next # 하나 저장했으니 다음 위치로 이동

    while len(arr) > 1:
        first = arr.pop(0) # 리스트 맨 처음꺼 추출(확인이 아니라 추출임)
        last = arr.pop()# 리스트 맨 마지막꺼 추출(확인이 아니라 추출임)
        if first != last: #만약 같지 않다?
            return False #실패~

    return True
```

```python
print(isPalindrome(l1))
```

```python
assert isPalindrome(l1) # 아무 결과가 안뜨면 통과임 assert는 이 뒤의 값이 무조건 True를 반환해야하고 아니면 무조건 error를 뛰움
```

```python
isPalindrome(l2)
```

```python
assert not isPalindrome(l2) #assert는 그 반대
```
