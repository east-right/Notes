### 다익스트라 값이 저장되어 있는 txt 불러와서 그래프화 시키기

    - sys.stdin.readlin == input() 이다. 아니 코딩으로 내가 짠게 아니라 진짜로 그 인수분해처럼 input()을 풀어쓰면 stdin저 줄들임
    - readline을 안쓰면 한번에 input("텍스트 내용")하는거랑 진배없는데 readline을 써서 줄바꿈 기준으로 input한다고 보면 된다. 
    - 이제 첫줄을 받는데 첫 줄은 그래프에 대한 설명이다. '5 10' 이렇게 있는게 split()로 ['5','10']인데 map()으로 인해 모든 값에 int()함수가 적용된다.
    - 그래서 5와 10이 나온다.
    - txt의 두 번째 줄을 받는다. 그게 1인데 작성자가 2번째 줄은 다익스트라 알고리즘할 때 시작 노드를 가르킨다고 했다.
    - 이제 graph를 그릴 도화지를 준비한다. 그래프 노드 개수보다 +1 한 이유는 노드 이름을 숫자로 할건데 0은 없으니까
    - 이제 차례대로 그래프에 노드와 엣지를 그릴건데 예시로 처음으로 들어오는 값을 보면 1 2 10이다.
    - 이건 1번 노드에서 2번 노드로 가는 단방향 엣지의 가중치(value)가 10이란 뜻이다.
    - 2 3 1는 2번 노드에서 3번 노드로 가는 단방향 엣지의 가중치가 1이란 뜻이다. 이렇게 쭉쭉 넣는다.
    - 처음 값은 노드를 나타내는거니 리스트의 index로 들어가야하고 엣지의 방향과 가중치는 튜플로 저장한다.
- 두 번째 쉘
    - 그럼 graph 결과가 나오는데 0번 노드는 없으니까 패스, index1에 있는 두 튜플은 1번 노드가 존재하는데 엣지를 두개 들고있다.
    - 각각 2번과 4번 노드를 가르키는 단방향 엣지이며 각각의 가중치는 10과 5이다.
    - 나머지도 위와 같이 해석하면 된다.

```python
import heapq


def dijkstra(graph, start):
    N = len(graph)
    INF = 1e9
    dist = [INF] * N

    # q로 최소 힙리스트를 만든다. 
    q = []
    
    # 인자값이 튜플일 경우 0번째 요소 기준으로 최소 힙 구조.
    # 첫 번째 방문 누적 비용은 0이다.
    heapq.heappush(q, (0, start))# heapq 메서드를 사용해서 리스트에 접근하면 리스트를 최소힙으로 받아들이면서 접근한다. 최대힙은 지원하지 않는다. 
    dist[start] = 0

    while q:
        # 누적 비용이 가장 작은 녀석을 꺼낸다.
        acc, cur = heapq.heappop(q)

        # 이미 답이 될 가망이 없다.
        if dist[cur] < acc:
            continue

        # 인접 노드를 차례대로 살펴보며 거리를 업데이트한다.
        for adj, d in graph[cur]:
            cost = acc + d
            if cost < dist[adj]:
                dist[adj] = cost
                heapq.heappush(q, (cost, adj))

    return dist
```

### 다익스트라 알고리즘 돌리는 코드
- 키워드 **최소힙**

- 그래프를 제시하고 다익스트라를 사용해서 알고 싶은 경로의 최초 시작점을 입력한다.
- 결과는 리스트 형태로 받을 것 이고 1e9는 그냥 엄청 큰 값을 줘서 일단 값이 무한정으로 변할 수 있다는걸 나타낸다.
- dist는 다익스트라 알고리즘을 타고 나온 1번에서 시작해서 해당 노드에 도착할 때 엣지의 가중치의 합이 최솟값(엣지의 가중치가 보통 거리라고 본다.)을 나온 값을 리스트로 반환해야 한다.
- 일단 초기는 전부 무한대로 저장
- q는 엣지를 통해 이동했을 때 해당 노드의 누적값을 기록한다. 기록 되는 방식은 (0,1) (5,4) 이렇게 기록된다.
- 튜플의 앞의 값은 노드의 누적 거리(엣지의 가중치를 합한거) 뒤의 값은 노드 번호(이름)이다.
- dist에는 1번 노드는 시작점이니 엣지를 탄적이 없다. 그래서 초기값 느낌의 0을 index 1에 넣는다.
- q 즉 엣지를 타고 도달했을 때 의 누적값에 대한 모든 정보를 확인할 때 까지 반복한다고 생각하면 된다.
- 일단 지금 들고있는 최소힙을 뺀다. 그 이유는 우리는 항상 누적값이 최소인 노드로 움직일 꺼니
- if는 백트래킹이라고 생각하면된다. 최단 엣지 가중치를 가질 희망이 없는 즉 이번 시점에서 이동해야할 노드의 누적값이 기존의 답안지보다 크면 글로는 이동할 이유가 없다. 그니까 그건 그냥 빼기만하고 괜히 거기로가서 정보를 탐색하지 마라는 것 이다
- for문이 다익스트라 엣지를 타는 코드이다.초기 힙리스트에 들어가서 pop으로 인해 나온 (0,1) 값의 기준으로 최초 for문이 돌아간다.
- cur은 1이고 acc는 지금 0이다. 여기서 for 문으로 인해 [(2, 10), (4, 5)]이 차례로 나온다. 즉 adj는 2고 d는 10이다.
- adj는 데이터 설명때 말했지만 2번 노드로 가는 엣지가 한개 존재한다는 건데 그 가중치는 10이란거다.
- 아니 근데 왜 헷갈리게 데이터에 1 2 10 이렇게 해서 (2,10)으로 튜플 들어가게 했냐. heapq가 튜플의 앞을 기준으로 최소힙 추출을 해주면 그냥 1 10 2로 하지
- 1 2 10으로 하는게 국룰이랍니다. 그냥 받아 들이고 작성해야함 ㅇㅇ
- cost를 계산하는데 이전 누적값(acc)과 이번에 엣지를 타면서 새로운 가중치(d)를 더해서 해당 노드에 도착했을 때 누적값을 가진다.
- 이제 여기서 dist 라는 최종 다익스트라에 2번 노드의 자리(index 2)에 있는 누적값(이전 다익스트라 루트로 이동하면서 해당 노드의 누적값 쌓았을 때 최솟값 이었던거)이번에 나온 누적값과 비교하고 적으면 이번 cost를 최솟값으로 다시 설정한다. 
- 이제 마지막으로 if문에 걸린다. == 이 엣지를 타는게 이번 노드로가는 최솟 코스트이다. 에 합리하여 이동한 노드의 엣지들을 확인하고 다음 노드들의 누적값을 확인한다.
- 이 작업을 반복하면 시작 노드로 부터 각 노드에 도달하기 위해 최소 거리(최소 엣지 가중치)를 모든 노드에서 알게 된다.

# 문제) 네트워크 딜레이 타임

주어진 노드 k에서 신호를 보내면 모든 노드가 신호를 수신하는 데 걸리는 최소 시간을 반환합니다. 모든 노드가 신호를 수신하는 것이 불가능하면 -1을 반환합니다.

Input: times = [[2,1,1],[2,3,1],[3,4,1]], n = 4, k = 2

Output: 2

Input: times = [[1,2,1]], n = 2, k = 1

Output: 1

```python
def delay_time(arr, n, k):
    graph = {}
    for start, end, time in arr:
        if start not in  graph:
            graph[start] = []
        graph[start].append((end,time))

    q = [(0,k)]
    dist = {}

    while q:
        acc, cur = heapq.heappop(q)
        if cur not in dist:
            dist[cur] = acc
            if cur in graph:
                for adj, t in graph[cur]:
                    #heapq.heappush(q,(acc +t, adj)) # 아래와 같은거임
                    q.append((acc +t, adj)) # 위와 같은거임
    if len(dist) == n:
        return max(dist.values())
    
    return -1
```

```python
def delay_time(arr, n, k):
    graph = {}

    for u, v, w in arr:
        if u not in graph:
            graph[u] = []
        graph[u].append((v, w))

    Q = [(0, k)]
    dist = {}

    while Q:
        time, node = heapq.heappop(Q)
        if node not in dist:
            dist[node] = time
            if node in graph:
                for v, w in graph[node]:
                    alt = time + w
                    heapq.heappush(Q, (alt, v))

    if len(dist) == n:
        return max(dist.values())

    return -1
```
