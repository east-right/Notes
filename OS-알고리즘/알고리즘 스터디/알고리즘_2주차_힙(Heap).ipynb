# 최대힙 구현

```python
class BinaryMaxHeap():
    def __init__(self): 
        self.items = [None] # 힙트리는 트리의 개념이 Node일지라도 보통 배열 형식으로 구현한다. 난이도 하락 위래 Index 0은 쓰지 않게 첫 값은 None으로 채워준다.
        
    def insert(self, val): # 새값을 넣기
        self.appned(val) # 새값은 라스트의 맨 뒤에 넣는다.
        
        cur = len(self.items) - 1 # 제일 마지막 인덱스를 구한다. // 인덱스는 원래 0부터 시작하니깐 길이 -1 을 해줘야 마지막 인덱스 이다.
        par = cur //2 # 완전 이진트리에서 부모 노드의 인덱스를 구하는 방법은 2의 몫을 구하는 것 이다.
        
        while par >0: # 아래의 작업을 RootNode까지 반복한다.
            if self.items[cur] > self.items[par]: #val이 부모 노드인 par인덱스의 값이 보다 크면
                self.items[cur], self.items[par] = self.items[par], self.items[cur] #val을 par 인덱스에 넣고, 기존 par 인덱스에 있는 값이 cur 인덱스로 보내진다.
                # 1. if절이 통과 되면 val이 par 인덱스에 들어감 2. 만약 if절을 타지 않았으면 val은 여전히 cur에 머물러 있다. 
            
            cur = par # 지금 상태의 par을 현 시점의 cur로 저장한다.
            par = cur // 2 #레벨이 올랐으니 다시 부모노드의 위치를 찾아야 하니 찾는다.
            
            """
            해당 작업은 새로들어온 val이 실제로 값이 제일 작아도 Rootnode 까지 반복을 한다.
            """
            
    def extract(self):
        """
        힙의 추출은 항상 Root 값을 반환하고, Root를 삭제시킨 다음 다시 힙의 조건에 맞춰 트리를 재 수정해야한다.
        """
        if len(self.items) < 1: # 비어있는 트리 행렬일수도 있으니까
            return None
        
        root = self.items[1] #index 0은 무조건 None이니 인덱스1이 RootNode이다.
        self.items[1], self.items[-1] = self.items[-1], self.items[1] # 추출 알고리즘은 root를 제일 뒤로 보내고 제일 뒤에거를 root로 보낸다.
        self.items.pop() #추출 // 지금 생각해보니간 root 값을 배정하는게 아니라 이 값을 root로 해도될듯 ㅇㅇ
        
        self._percolate_down(1) #이제 가장 큰 값이 빠졋으니 기존 leafnode가 Root에 가 있으니 작업을 해준다.
        
        return root
        
        
    def _percolate_down(self, cur):
        largest = cur # 이 함수의 주력 //초기값 기준 인덱스 1이 들어옴 // 초기 값이 아니면 뭐가 들어올지 모름 //부모노드를 일단 주력으로 만듬
        left = cur *2
        right = cur *2 +1
        
        if self.items[left] > self.items[largest]: #부모 보다 왼쪽 아들이 값이 더 크면 기록록
            largest = left
            
        if self.items[right] > self.items[largest]: # 부모 보다 오른쪽 아들이 값이 더 크면 기록록
            largest = right
            
        if largest != cur: # 위의 if 문 두 개중 하나라도 걸렸으면 여기서도 걸린다.
            self.items[largest], self.items[cur] = self.items[cur], self.items[largest] #가장 큰 값이랑 
            self._percolate_down(largest) #걸렸으면 다시 무한 츠쿠요미 마냥 한번더 반복해라
```

```python
# 이건 강사님이 작성한 코드드
class BinaryMaxHeap:
    def __init__(self):
        # 계산 편의를 위해 0이 아닌 1번째 인덱스부터 사용한다.
        self.items = [None]

    def insert(self, k):
        self.items.append(k)
        self._percolate_up()

    def _percolate_up(self):
        # percolate: 스며들다.
        cur = len(self.items) - 1
        # left 라면 2*cur, right 라면 2*cur + 1 이므로 parent 는 항상 cur // 2
        parent = cur // 2

        while parent > 0:
            if self.items[cur] > self.items[parent]:
                self.items[cur], self.items[parent] = self.items[parent], self.items[cur]

            cur = parent
            parent = cur // 2

    def extract(self):
        if len(self.items) < 2: 
            return None

        root = self.items[1]
        self.items[1] = self.items[-1]
        self.items.pop()
        self._percolate_down(1)

        return root

    def _percolate_down(self, cur):
        biggest = cur
        left = 2 * cur
        right = 2 * cur + 1

        if left <= len(self.items) - 1 and self.items[left] > self.items[biggest]:
            biggest = left

        if right <= len(self.items) - 1 and self.items[right] > self.items[biggest]:
            biggest = right

        if biggest != cur:
            self.items[cur], self.items[biggest] = self.items[biggest], self.items[cur]
            self._percolate_down(biggest)
```
