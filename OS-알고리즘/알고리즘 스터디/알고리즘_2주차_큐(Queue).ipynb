# Queue 기초

```python
class Node:
    def __init__(self, val, next=None): #기본적으로 Node에는 값(Value)와 다음 위치(next)를 들고 있어야 한다.
         self.val = val
         self.next = next
```

```python
"""
코드를 보면 알겠지만 pop 빼곤 LinkedList와 하나도 빠짐 없이 같은 코드를 취하고 있다. 즉 LinkedList를 이해하고 있으면 이 코드는 어려울게 없다.
"""

class Queue:
    def __init__(self):
        self.front = None # 호출시 비어있음
        
    def push(self, val):
        if not self.front: # front node가 없으면 = 첫 Node를 집어 넣는거면
            self.front = Node(val) # front로 설정정
            return
        
        node = self.front #그게 아니면 front node에서 부터
        while node.next: #None 값이 나올 때 까지 반복
            node = node.next # node.next에 있는 Node를 node로 지정 = 계속 타고 내려감
        node.next = Node(val) #거기에 Node를 넣어라라
    
    def pop(self):
        if not self.front: 
            return None # 비어있으면 None
        
        node = self.front #front -> node로 선택 -> 이건 값을 보여주기 위해 node로 지정
        self.front = node.next # 현 front node의 next를 이제 front로 지정
        return node.val # 현 front 였던 node의 값을 보여주고 장렬히 전사
    
    def is_empty(self):
        return self.front is None
```

```python
def test_queue():
    queue = Queue()

    queue.push(1)
    queue.push(2)
    queue.push(3)
    queue.push(4)
    queue.push(5)

    assert queue.pop() == 1
    assert queue.pop() == 2
    assert queue.pop() == 3
    assert queue.pop() == 4
    assert queue.pop() == 5
    assert queue.pop() is None
    assert queue.is_empty()
```

```python
test_queue()
```

```python
#강의에선 deque라는걸 사용해서 간단하게 풀 수 있다고 한다. 자세한건 내 공부 markdown
from collections import deque

def test_problem_queue(num):
    deq = deque([i for i in range(1, num + 1)])
    while len(deq) > 1:
        deq.popleft()
        deq.append(deq.popleft())
    return deq.popleft()
```

```python
# 하지만 일단 큐로만 구현이 가능하기에 직접해보자(은근 오래 걸렷다. is_empty의 존재 의의를 완벽히 알았다.)
def code(lst):
    queue = Queue()
    for i in lst:
        queue.push(i) # 입력된 숫자 리스트 일단 큐로 다 하나씩 집어 넣음음
        
    val_first = None 
    val = None
    
    while not queue.is_empty(): # while은 해당 조건이 False가 나올 때까지 아래 코드를 반복한다. 그래서 비어있지(None) 않으면 이 False가 나오는 is_empty()를 사용함 
        val_first = queue.pop() #일단 지시사항 대로 먼저 pop으로 뺀다.
        if queue.is_empty(): # 빼고 나서 비어있는지 확인한다. 비어있으면 True가 나오기에 무한루프 종료
            break
        
        val = queue.pop() # 지시사항대로 두번째는 뒤로 옮겨야 한다 => 삭제후 추가 작업을 해야한다. 요기선 삭제
        queue.push(val)# 요기선 추가
    return val_first #그렇게 해서 첫 pop을 하고나니 is_empty에 걸리는 val_first을 반환
```
